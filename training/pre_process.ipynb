最适合使用的数据集是content表中的标注结果，文本长度差异小，同时各种情感都有足够的数据量
可以将其进行预处理，用于后面的训练。

```python
length=data['内容'].astype(str).str.len()
length_count = length.value_counts()
```

```python
# 定义分箱的边界
bins = list(range(0, max(length) + 100, 100))
labels = [f'{bins[i]}-{bins[i + 1]}' for i in range(len(bins) - 1)]

# 对文本长度进行分箱
length_bins = pd.cut(length, bins=bins, labels=labels, right=False)

# 统计每个分箱的数量
bin_counts = length_bins.value_counts().sort_index()
```

```python
# 绘制饼图
plt.plot(labels,bin_counts)
plt.title('文本长度分布折线')
```

可以看到超过300字的数据非常少，接近于0，可以去掉，对后续lstm的字数限制进行处理，同时切开的句子有部分过于短，也需要去掉

下面平均各个标签的数据数量，删除neutral 与 negative 中超过 positive 部分的数据

```python
positive = len(data[data['emotions'] == 'positive'])
positive
```

```python
new_data = pd.DataFrame()
```

```python
#对于超过的部分进行去除，随机的选出两个标签中的数据
# 使用groupby实现
random_seed = 42
balanced_data = data.groupby('emotions', group_keys=False).apply(
    lambda x: x.sample(n=positive, random_state=random_seed) 
    if x.name in ['negative', 'neutral'] 
    else x
)
```

```python
balanced_data
```

```python
balanced_data['emotions'].value_counts()
```

对于处理好的训练数据，进行存储，至此的数据可以正式用于训练

```python
#存入数据库的表
cursor=connect.cursor()
cursor.execute('''
CREATE TABLE IF NOT EXISTS data(
    id integer primary key autoincrement,
    content TEXT,
    word TEXT,
    emotion TEXT,
    score double
)
''')
```

```python
balanced_data.to_sql('data',connect, if_exists='replace', index=False)
```
